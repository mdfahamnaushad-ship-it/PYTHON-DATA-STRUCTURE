#Q1 - What are data structures, and why are they important?
  - Data structures are special ways of organizing and storing data so that we can use it efficiently.
  They are important because they make it easier to manage and access data quickly. For example, lists store items in order, dictionaries use key-value pairs, and sets avoid duplicates.

#Q2 - Explain the difference between mutable and immutable data types with examples.

  - Mutable data types can be changed after they are created, while immutable ones cannot.

    Example:

    Mutable: List → a = [1, 2, 3]; we can do a[0] = 10.
  
    Immutable: Tuple → b = (1, 2, 3); trying b[0] = 10 will give an error.

#Q3 - What are the main differences between lists and tuples in Python?
    | Feature    | List              | Tuple      |
    | ---------- | ----------------- | ---------- |
    | Mutability | Mutable           | Immutable  |
    | Syntax     | Uses `[]`         | Uses `()`  |
    | Speed      | Slower            | Faster     |
    | Use case   | Data that changes | Fixed data |

#Q4 - Describe how dictionaries store data.
  - Dictionaries store data in key-value pairs. Each key must be unique, and it maps to a specific value.

    Example: {"name": "FAHAM", "age": 29} → "name" is the key, "FAHAM" is its value.

#Q5 - Why might you use a set instead of a list in Python?
  - A set is used when we want unique elements and don't care about order.
    They are faster for checking if an item exists in a collection.

#Q6 - What is a string in Python, and how is it different from a list?
  - A string is a sequence of characters enclosed in quotes.

    It looks similar to a list but is immutable — you can't change characters inside it directly, while list elements can be changed.

#Q7 - How do tuples ensure data integrity in Python?
  - Tuples are immutable, so once created, their values cannot be modified or deleted. This makes them safer when you need to store constant data.

#Q8 - What is a hash table, and how does it relate to dictionaries in Python?
  - A hash table is a data structure that stores data in key-value form and   uses a hash function to quickly find keys.

    Python's dictionaries are built on hash tables internally.

#Q9 - Can lists contain different data types in Python?
  - Yes, a single list can store elements of different types.

    Example: [10, "Hello", 3.5, True]

#Q10 - Explain why strings are immutable in Python.
  - Strings are immutable for safety and performance reasons. Changing a string would require creating a new one, which avoids unwanted side effects in the program.

#Q11 - What advantages do dictionaries offer over lists for certain tasks?
  - Dictionaries provide fast lookups using keys, whereas lists require searching through all elements.

    For example, finding a value by name in a dictionary is much faster.

#Q12 - Describe a scenario where using a tuple would be preferable over a list.
  - When you have data that shouldn't change, like coordinates (x, y) or days of the week.

    Tuples prevent accidental modifications.

#Q13 - How do sets handle duplicate values in Python?
  - Sets automatically remove duplicates.

    Example: {1, 2, 2, 3} becomes {1, 2, 3}.

#Q14 - How does the “in” keyword work differently for lists and dictionaries?
  - *In a list, it checks if an element exists.

 -  *In a dictionary, it checks if a key exists.

     Example:-



```python
3 in [1, 2, 3]  # True
"name" in {"name": "Faham", "age": 29}  # True
```

#Q15 - Can you modify the elements of a tuple? Explain why or why not.
  - No, you can’t modify tuple elements because tuples are immutable. Once created, their contents stay fixed.

#Q16 - What is a nested dictionary, and give an example of its use case?
  - A nested dictionary is simply a dictionary that contains another dictionary as one or more of its values.
    It’s useful when you want to store structured or grouped data — like data that belongs together
  
    Example:

```python
student = {
    "name": "Faham",
    "age": 29,
    "marks": {
        "math": 85,
        "science": 90,
        "english": 88
    }
}

print(student["marks"]["science"])
```

#Q17 - Describe the time complexity of accessing elements in a dictionary.

  - Accessing elements by key in a dictionary takes O(1) time on average because of hashing.

#Q18 - In what situations are lists preferred over dictionaries?

 -  When data doesn’t need key-value mapping or order matters, like a sequence of numbers or names.

#Q19 - Why are dictionaries considered unordered, and how does that affect data retrieval?

  - Earlier versions of Python didn’t guarantee order in dictionaries.

    Even though modern versions maintain insertion order, retrieval is still based on keys, not positions.

#Q20 - Explain the difference between a list and a dictionary in terms of data retrieval.

  - List: Accessed by index → my_list[0]

  - Dictionary: Accessed by key → my_dict["name"]
